```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

print("All libraries imported successfully!")
```

```python
# Initialize the Chrome driver automatically
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

print("Browser initialized successfully!")
```

```python
# Tell the browser to open a webpage
url = "https://quotes.toscrape.com/"
driver.get(url)

print("Successfully navigated to the webpage!")
```

```python
# Find elements on the page (for example, looking for quote text on the page)
# We will grab all quote blocks
quote_elements = driver.find_elements(By.CSS_SELECTOR, ".quote")

extracted_data = []

for el in quote_elements:
    try:
        # Extract the text of the quote and the author
        text = el.find_element(By.CSS_SELECTOR, ".text").text
        author = el.find_element(By.CSS_SELECTOR, ".author").text
        
        extracted_data.append({
            "Quote": text,
            "Author": author
        })
    except Exception as e:
        continue

print(f"Successfully extracted {len(extracted_data)} quotes!")
```

```python
# Convert the list into a Pandas DataFrame
df = pd.DataFrame(extracted_data)

# Display the first few rows nicely in Jupyter
display(df)

# Save the data to a CSV file
df.to_csv("extracted_quotes.csv", index=False)
print("Data successfully saved to 'extracted_quotes.csv'!")
```
